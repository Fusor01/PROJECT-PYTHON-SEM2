#importing dependencies


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split #random spliting of data by rows
from sklearn.feature_extraction.text import TfidfVectorizer #converting txt to umerical vectirs
from sklearn.linear_model import LogisticRegression # calling algorithms
from sklearn.metrics import accuracy_score #model evaluation
```

#data collection and pre porcessing

```python
# loding data from csv to pandans df
raw_mail_data =pd.read_csv("/content/mail_data.csv")
```

```python
print(raw_mail_data)
```

```python
# replacing the null values with null strings
mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)), '')
```

```python
#printing the 1st 5 rows
mail_data.head()
```

```python
#chekinf the ni of rows and coloumns in dataframe
mail_data.shape
```

```python
# lable encoding : replacing the lable value with a numerical value here for example we are lablelling ham to 1 and spam as 0

mail_data.loc[mail_data['Category']=='spam','Category',]=0
mail_data.loc[mail_data['Category']=='ham','Category',]=1
```

ham = 1 ; spam = 0

```python
mail_data.head()
```

spliting the data into features and targets

seprating the data into text(messages) and lables(type : spam  and ham)


```python

X = mail_data['Message']

Y = mail_data['Category']
```

```python
print(X)
```

```python
print(Y)
```

spliting the data into train ing and testing data

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y,test_size=0.2,random_state=6)
```

```python
print(X_test.shape)
print(X_train.shape)
print(X.shape)
```

to convert the textual date of the mesagees to meaningful numerical values:  aka  FEATURE EXTRACTION *(the proces of converion is vectorization}*

```python
# transforming the text data to feature vectors that can be used as input to the LOGISTIC REGRESSION
# gives the weight score to each repeated words and will be tied to ots label
#  min _df=1 :if min score of a word is less than 1 ignore it and if it is gereater than 1 then include it
# stop_words='english' : includes a set of common words that would be ignored from weighting like: is, am ,was,,etc for a specified laguage here english
# lowerxase = 'True' : converts al the text to lower case for bwtter processing
# basicaly to my crux this fuction is the tokenizer + weighter and


feature_extraction = TfidfVectorizer(min_df = 1 , stop_words = 'english', lowercase=True)
# loaded tfidfvectorizer to a variavle feature_extracton
X_train_features = feature_extraction.fit_transform(X_train) #fit data to the vectorixer fuction
X_test_features = feature_extraction.transform(X_test) # why?

# convert Y_train and Y_test as integers

Y_train = Y_train.astype("int")
Y_test = Y_test.astype("int")
```

```python
print(X_train_features)
```

```python
print(X_test_features)
```

#TRAINING THE MACHINE LEARNING MODEL

LOGISTIC REGRESSION


```python
#loding the regression model to the variable model
model = LogisticRegression()
```

```python
# training the logistic regression model with the trainign data
model.fit(X_train_features,Y_train)
```

# EVALUATING THE MODEL

1ST WE ARE TESTING THE MODEL ON TRAINIG DATA ITSELF

```python
#PREDICTION ON TRAINING DATA  (here basically the model will try to evsluate wheter the mail is - 0 or 1 ie spam or ham mail respectively)

prediction_on_training_data = model.predict(X_train_features)

# bASICALLY WE TRY TO COMPARE THE ACTUAL VALUE OF 0 AND 1 TO WHAT THE MODEL PREDICTED AND GIVE US A SCORE in between 0 and 1
accuracy_on_training_data = accuracy_score(Y_train , prediction_on_training_data )
```

```python
print(accuracy_on_training_data)
```

```python
print(f'ACCURACY ON TRAINING DATA :{accuracy_on_training_data*100}%')
```

2ND WE ARE TESTING WITH TEST DATA

```python
#PREDICTION ON Testing DATA  (here basically the model will try to evsluate wheter the mail is - 0 or 1 ie spam or ham mail respectively)

prediction_on_testing_data = model.predict(X_test_features)

# bASICALLY WE TRY TO COMPARE THE ACTUAL VALUE OF 0 AND 1 TO WHAT THE MODEL PREDICTED AND GIVE US A SCORE in between 0 and 1
accuracy_on_testing_data = accuracy_score(Y_test , prediction_on_testing_data )
```

```python
print(accuracy_on_testing_data)
```

```python
print(f'ACCURACY ON TESTING DATA :{accuracy_on_testing_data*100}%')
```

the reason of using traingin data for testigna 1st is the case of overfitting and under fitting

overfitting:model performs verywell on training data fut performs poorly on testing data

under fitting : model performs overwhelming on traing dATA but performs very well on testing data

***but for this to be true the diffrece between the acuracy must be significant loke 98 and 60 not 96 and 97***
